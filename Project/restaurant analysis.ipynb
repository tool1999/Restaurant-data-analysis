------------------------  Cleaning Stage------------------
                                                       

```python
#1.We will start by the payment method col 
df['Payment Method'].nunique()
```

```python
df['Payment Method'].unique()
```

```python
df['Payment Method'].value_counts()
```

```python
#there are nan values so we need to know is it missing or what
df['Payment Method'].isnull().mean().round(2)*100 
#We will fill this 6% with unknown value till making a relation betwwen al nulls col. so if tere are any raw has nulls we will drop it 
```

```python
df['Payment Method']=df['Payment Method'].fillna('unknown')
```

```python
df['Payment Method'].unique()
#now we have 4 payment methods
```

```python
#converting data type of col Order date from str to datetime
df['Order Date']=pd.to_datetime(df['Order Date'])
```

```python
df.info()
```

```python
df.head()
#now we destingutish that Order total is the result of multiplication of price with quantity 
#so we can  1.Rename Order Total to be Total price
           #2.calculating the missing values of any col.of these three col. from the athor two col.  
```

```python
#1.Changing col.name
df.rename(columns={'Order Total':'Total price'},inplace=True)
```

```python
df.head()
```

```python
df.isna().mean().round(2)*100
#We distigutish that  price missing value is 5% & quantity & total price 2% + price is result of division of total price divided by quantity 
```

```python
#here we mask the price col. has null values when the Quantity & Total price have values
mask=(df['Price'].isna()&
df['Quantity'].notnull() &
df['Total price'].notnull())
df[mask]
# We will calculate the 446 rows of price col. from the another two col.
```

```python
df.loc[mask,'Price']=(df.loc[mask,'Total price'] / df.loc[mask,'Quantity'])
```

```python
df[mask].head(3)
```

```python
df.isna().mean().round(2)*100
#now we found that % of missing value of price decreased to 2%
```

```python
#Now We can make deep cleaning in the three col. which have 2% missing values 
# if the three col. have the missing values we can drop them 
mask2=(df['Price'].isnull()&
       df['Quantity'].isnull() &
       df['Total price'].isnull()
       &df['Item'].isnull())
df[mask2]
# So now it is clean that this 430 rows are missing values no data in col "item , price , quantity , total price & some of payment method are unknown "
# So we can drop this rows to clean our data as it acts only 2% 
```

```python
#Drop missing data in col item , price , quantity ,total price 
df.dropna(subset=['Item','Price','Quantity','Total price'],how='all',inplace=True)
```

```python
df.isna().mean().round(2)*100
#Now the item missing values decreased to 8 % & we can analys according to Category of items
```

```python
#some payment method were unknown & others are deleted with the nan values of the prevous col
# we need to know ho many known payment method
df['Payment Method'].value_counts(normalize=True)*100
```

                                               Now we can starting analysis of The Data set

                                                    A.Customer column analysis

1.What are the top 10 customers orders?

```python
cust_sale=df.groupby('Customer ID')['Order ID'].count().reset_index().sort_values(ascending=False,by='Order ID').head(10)
px.bar(cust_sale,x='Customer ID',y='Order ID',text_auto=True,title='Top Ten no.of orders')
```

2.Top customers payment

```python
cust=df.groupby('Customer ID')['Total price'].sum().reset_index().sort_values(ascending=False,by='Total price').head(10)
px.histogram(cust,x='Customer ID',y='Total price',text_auto=True,title='Top Ten Payment')
```

3.least 5 orders

```python
cust1=df.groupby('Customer ID')['Total price'].count().reset_index().sort_values(ascending=False,by='Total price').tail(5)
px.histogram(cust1,x='Customer ID',y='Total price',text_auto=True)
```

4.How many times each payment method used

```python
paymeth=df.groupby('Payment Method')['Customer ID'].count().reset_index()
px.bar(paymeth,x='Payment Method',y='Customer ID',title='presentage of payment method',text_auto=True).update_xaxes(categoryorder='total descending')
```

5.How many customers in each month

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

cust_mon = (
    df.groupby(df['Order Date'].dt.to_period('M'))['Customer ID']
      .nunique()      # عدد العملاء المختلفين في كل شهر
      .reset_index()
)

cust_mon['Order Date'] = cust_mon['Order Date'].astype(str)

px.bar(cust_mon, x='Order Date', y='Customer ID',text_auto=True,title='Customers in each month')
```

6.How many Orderes of each items

```python
ord_item=df.groupby('Item')['Customer ID'].count().reset_index()
px.bar(ord_item,x='Item',y='Customer ID',title='Amount of orders of items',text_auto=True,color_discrete_sequence=['green']).update_xaxes(categoryorder='total descending')
```

7.Presentage of each Dish

8.Average item price

```python
avg=df.groupby('Item')['Price'].mean().reset_index()
px.histogram(avg,x='Item',y='Price',title='Average cost of item',text_auto=True).update_xaxes(categoryorder='total descending')
```

9.Best selling Item

```python
sell=df.groupby('Item')['Total price'].sum().reset_index()
px.histogram(sell,x='Item',y='Total price',title='Best item income',text_auto=True).update_xaxes(categoryorder='total descending')
```

10.Quantity of each item

```python
#we made reset index avoid counting item as row so the index of rows starts from pasta alfredo
result=df.groupby('Item')['Quantity'].sum().reset_index().sort_values(ascending=False,by='Quantity')
px.bar(result,x='Item',y='Quantity',text_auto=True,title='Best selling Items',color_discrete_sequence=['yellow'])
```

                                                        C.Date time col

11.Sales of each year

```python
sales_by_year = (df.groupby(df["Order Date"].dt.year)["Total price"].sum().reset_index())

px.bar(sales_by_year,x='Order Date',y='Total price',title='Sales of each year',text_auto=True)
```

12.Orders in months of year 2022

```python
sales_2022=df[df['Order Date'].dt.year == 2022]
Sa22=sales_2022.groupby(sales_2022['Order Date'].dt.month)['Total price'].sum().reset_index()
# We need to create a new col.named Month 
Sa22['Month']=pd.to_datetime(Sa22['Order Date'],format='%m').dt.month_name()

px.line(Sa22,x='Month',y='Total price',title='Sales of months of year 2022',labels=True,markers=True,)
```

13.Top selling day in 2022

```python
sales_2022 = df[df["Order Date"].dt.year == 2022]

best_day =sales_2022.groupby("Order Date", as_index=False)["Total price"].sum().sort_values("Total price", ascending=False).reset_index()
best_day.head(1)
```

14.Top selling day in 2023

```python
sales_2023 = df[df["Order Date"].dt.year == 2023]

best_day =sales_2023.groupby("Order Date", as_index=False)["Total price"].sum().sort_values("Total price", ascending=False).reset_index()
best_day.head(1)
```
